```julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    T = zeros(Expression, 27, 27)

    for k in 1:23
        U_k, V_k, W_k = a[k]
        T += kron(kron(U_k, V_k), W_k)
    end

    T .= simplify_expr.(T)
    return T
end

T = compute_T(lad)
```

```julia
# The 3 x 3 matrix-multiplication tensor.
mm = 3; nn = 3; pp = 3

tenset = Matrix{Int}[]

for i in 1:mm, j in 1:nn, k in 1:pp
    matrix1 = zeros(Int, mm, nn)
    matrix2 = zeros(Int, nn, pp)
    matrix3 = zeros(Int, pp, mm)

    matrix1[i,j] = 1
    matrix2[j,k] = 1
    matrix3[k,i] = 1

    push!(tenset, kron(kron(matrix1, matrix2), matrix3))
end

T1 = sum(tenset)
```
