## Six-parameter Laderman family `lad_t6`

This notebook starts from the numerical Laderman decomposition in `lad(1).ipynb` and substitutes the solution obtained from a Gröbner basis.

The parameter vector is
$$
t=(t_1,t_2,t_3,t_4,t_5,t_6)\in(\mathbb C^*)^6,
$$
where
$$
\begin{aligned}
t_1&=x[1,1,4], & t_2&=z[2,1,4], & t_3&=z[1,2,4],\\
t_4&=x[1,1,6], & t_5&=x[2,2,16],& t_6&=z[1,2,16].
\end{aligned}
$$

All entries not belonging to the selected 47 coordinates remain equal to their original numerical values. The original Laderman point is recovered at
$$
(t_1,t_2,t_3,t_4,t_5,t_6)=(-1,-1,1,1,1,1).
$$

```julia
# Collections of the 23 factor matrices.
U = [U1, U2, U3, U4, U5, U6, U7, U8, U9, U10, U11, U12, U13, U14, U15, U16, U17, U18, U19, U20, U21, U22, U23]
V = [V1, V2, V3, V4, V5, V6, V7, V8, V9, V10, V11, V12, V13, V14, V15, V16, V17, V18, V19, V20, V21, V22, V23]
W = [W1, W2, W3, W4, W5, W6, W7, W8, W9, W10, W11, W12, W13, W14, W15, W16, W17, W18, W19, W20, W21, W22, W23]
```

```julia
# The 23 rank-one summands (U_k, V_k, W_k).
lad = [(U[k], V[k], W[k]) for k in 1:23]
```

```julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    T = zeros(Expression, 27, 27)

    for k in 1:23
        U, V, W = a[k]
        T += kron(kron(U, V), W)
    end

    T .= simplify_expr.(T)
    return T
end

T = compute_T(lad)
```

```julia
mm = 3; nn = 3; pp = 3

indices = [(i, j, k) for i in 1:mm for j in 1:nn for k in 1:pp]
tenset = []

for (i, j, k) in indices
    matrix = zeros(Int, mm, nn)
    matrix[i,j] = 1
    matrix1 = matrix

    matrix = zeros(Int, nn, pp)
    matrix[j,k] = 1
    matrix2 = matrix

    matrix = zeros(Int, pp, mm)
    matrix[k,i] = 1
    matrix3 = matrix

    tensor_product = kron(kron(matrix1, matrix2), matrix3)
    push!(tenset, tensor_product)
end

T1 = sum(tenset)
```
