```julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    b = zeros(Expression, 64, 64)

    for k in 1:48
        U, V, W = a[k]
        b += kron(kron(U, V), W)
    end

    b .= simplify_expr.(b)
    return b
end
```
