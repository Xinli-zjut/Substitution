## Two-parameter family `s722_t2`

This notebook starts from the numerical rank-23 decomposition in `s722.ipynb` and substitutes the solutions obtained from the Gröbner bases.

Choose $t_1=z[3,3,15]$ and $t_2=z[3,3,22]$.
The first Gröbner basis gives
$$
\begin{aligned}
z[1,1,20]&=t_1^{-1},&
z[3,3,8]&=t_1,&
y[3,3,21]&=t_1,\\
y[1,1,14]&=-t_1,&
y[3,3,13]&=-t_1^{-1},&
y[1,1,7]&=-t_1^{-1},\\
y[1,1,1]&=-t_1^{-1},&
x[1,1,20]&=-t_1,&
x[3,3,15]&=t_1^{-1},\\
x[3,3,8]&=-t_1^{-1},&
x[2,1,7]&=-t_1.
\end{aligned}
$$

The second Gröbner basis gives
$$
\begin{aligned}
z[2,3,22]&=-t_2,&
z[3,1,22]&=-t_2,&
z[2,1,22]&=t_2,\\
x[2,2,5]&=t_2,&
x[2,1,5]&=t_2.
\end{aligned}
$$

All other entries remain equal to their numerical values in `s722.ipynb`. The original decomposition is recovered at
$(t_1,t_2)=(-1,0)$.

```julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    U_1, V_1, W_1 = a[1]
    T = kron(kron(U_1, V_1), W_1)

    for k in 2:length(a)
        U_k, V_k, W_k = a[k]
        T = T + kron(kron(U_k, V_k), W_k)
    end

    return simplify_expr.(T)
end

T = compute_T(s722_t2)
```

```julia
# The 3 x 3 matrix-multiplication tensor.
mm = 3; nn = 3; pp = 3
T1 = zeros(Int, 27, 27)

for i in 1:mm, j in 1:nn, k in 1:pp
    matrix1 = zeros(Int, mm, nn)
    matrix2 = zeros(Int, nn, pp)
    matrix3 = zeros(Int, pp, mm)

    matrix1[i,j] = 1
    matrix2[j,k] = 1
    matrix3[k,i] = 1

    T1 = T1 + kron(kron(matrix1, matrix2), matrix3)
end

T1
```

```julia
# Exact symbolic verification of all 729 Brent equations.
D = simplify_expr.(T .- T1)
nonzero_positions = findall(q -> !iszero(q), D)

println("number of nonzero entries in T-T1 = ", length(nonzero_positions))
@assert isempty(nonzero_positions)
```
