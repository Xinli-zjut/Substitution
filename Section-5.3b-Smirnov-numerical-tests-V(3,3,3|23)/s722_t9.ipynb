## Nine-parameter family `s722_t9`

This notebook starts from the numerical rank-23 decomposition in `s722.ipynb` and substitutes a nine-parameter component obtained from a Gröbner basis.

Choose the free coordinates
$
\begin{aligned}
t_1&=z[3,3,8], & t_2&=z[3,3,22], & t_3&=z[1,1,11],\\
t_4&=z[1,1,23],& t_5&=z[1,2,1],  & t_6&=z[1,2,2],\\
t_7&=y[1,1,7], & t_8&=z[1,2,8],  & t_9&=y[1,3,16].
\end{aligned}
$

The formulas contain the denominators $t_1,t_4,t_5,t_6,t_7$, so the present affine chart is defined by
$t_1t_4t_5t_6t_7\ne0.$ The remaining four parameters $t_2,t_3,t_8,t_9$ may vanish.

The Gröbner basis is reducible at the original point.  Here we select the component determined, in particular, by
$$z[3,2,5]=z[2,3,23]=z[3,3,7]=y[3,2,7]=y[3,1,7]=y[3,3,4]=0,$$
while retaining $z[3,3,22]=t_2$ and $y[1,3,16]=t_9$ as free coordinates.

All parameter expressions are inserted directly into the corresponding matrices $(U_k,V_k,W_k)$. All entries not involved in the Gröbner-basis subsystem retain their numerical values from `s722.ipynb`.
The original `s722` decomposition is recovered at $(t_1,t_2,t_3,t_4,t_5,t_6,t_7,t_8,t_9)=(-1,0,0,1,-1,1,1,0,0)$.

```julia
function simplify_expr(e::Expression)
    e = expand(e)
    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)
    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    U1, V1, W1 = a[1]
    T = kron(kron(U1, V1), W1)

    for k in 2:length(a)
        Uk, Vk, Wk = a[k]
        T = T + kron(kron(Uk, Vk), Wk)
    end

    return simplify_expr.(T)
end

T = compute_T(s722_t9)
```

```julia
# The 3 x 3 matrix-multiplication tensor.
mm = 3; nn = 3; pp = 3
T1 = zeros(Int, 27, 27)

for i in 1:mm, j in 1:nn, k in 1:pp
    matrix1 = zeros(Int, mm, nn)
    matrix2 = zeros(Int, nn, pp)
    matrix3 = zeros(Int, pp, mm)

    matrix1[i,j] = 1
    matrix2[j,k] = 1
    matrix3[k,i] = 1

    T1 = T1 + kron(kron(matrix1, matrix2), matrix3)
end

T1
```

```julia
# Exact symbolic verification of all 729 Brent equations.
D = simplify_expr.(T .- T1)
nonzero_positions = findall(q -> !iszero(q), D)

println("number of nonzero entries in T-T1 = ", length(nonzero_positions))
@assert isempty(nonzero_positions)

nonzero_positions
```
