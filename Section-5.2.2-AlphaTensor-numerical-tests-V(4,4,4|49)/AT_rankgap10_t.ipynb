````julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
````

````julia
function compute_T(a)
    T = zeros(Expression, 64, 64)

    for k in 1:49
        U, V, W = a[k]
        T += kron(kron(U, V), W)
    end

    T .= simplify_expr.(T)
    return T
end

T = compute_T(sol526)
````

````julia
mm = 4; nn = 4; pp = 4

indices = [(i, j, k) for i in 1:mm for j in 1:nn for k in 1:pp]
tenset = []

for (i, j, k) in indices
    matrix = zeros(Int, mm, nn)
    matrix[i,j] = 1
    matrix1 = matrix

    matrix = zeros(Int, nn, pp)
    matrix[j,k] = 1
    matrix2 = matrix

    matrix = zeros(Int, pp, mm)
    matrix[k,i] = 1
    matrix3 = matrix

    tensor_product = kron(kron(matrix1, matrix2), matrix3)
    push!(tenset, tensor_product)
end

T1 = sum(tenset)
````

````julia
# The tensor identity is imposed on the hypersurface
#     t5*(t6+t7) - 1 = 0.
Delta = simplify_expr.(T - T1)
````
