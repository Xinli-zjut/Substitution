```julia
using Oscar
```

```julia
R, x, y, z = polynomial_ring(
    QQ,
    :x => (1:4, 1:4, 1:49),
    :y => (1:4, 1:4, 1:49),
    :z => (1:4, 1:4, 1:49)
)
```

```julia
Brentsub=include("Brent_unique_subsystem_1_gap10.jl")
```

```julia
Svars
```

```julia
I = ideal(R, Brentsub)
```

```julia
@time G=groebner_basis(I, ordering=lex(R))
```

```julia
#@time G=groebner_basis(I)     # Under the default ordering, the Gröbner basis is easier to compute.
                              # In many cases, the corresponding solution set can also be determined.
```

```julia
D=dim(I)
```

```julia
n_total = length(x) + length(y) + length(z)
```

```julia
n_S = length(Svars)
```

```julia
D_big = dim(I)
```

```julia
D_small = D_big - (n_total - n_S)
```

```julia
println("total variables = ", n_total)
println("#S variables = ", n_S)
println("free unused variables = ", n_total - n_S)
println("dim in full ring = ", D_big)
println("dim of small system = ", D_small)
```

```julia
open("groebner_basis_subs_1.txt", "w") do io
    println(io, "Groebner basis")
    println(io, "Number of elements = ", length(G))
    println(io, "--------------------------------------------------")

    for i in 1:length(G)
        println(io, "G[$i] = ", G[i])
        println(io)
    end
end
```
