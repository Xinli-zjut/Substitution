```julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    T = zeros(Expression, 64, 64)

    for k in 1:49
        U, V, W = a[k]
        T += kron(kron(U, V), W)
    end

    T .= simplify_expr.(T)
    return T
end

T = compute_T(sol721)
```
