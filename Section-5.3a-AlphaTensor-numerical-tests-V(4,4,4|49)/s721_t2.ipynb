```julia
function simplify_expr(e::Expression)
    e = expand(e)

    p, q = get_num_den(e)
    p = expand(p)
    q = expand(q)

    return isone(q) ? p : expand(p / q)
end
```

```julia
function compute_T(a)
    U_1, V_1, W_1 = a[1]
    T = kron(kron(U_1, V_1), W_1)

    for k in 2:length(a)
        U_k, V_k, W_k = a[k]
        T = T + kron(kron(U_k, V_k), W_k)
    end

    return simplify_expr.(T)
end

T = compute_T(s721_t2)
```

```julia
# The 4 x 4 matrix-multiplication tensor.
mm = 4; nn = 4; pp = 4
T1 = zeros(Int, 64, 64)

for i in 1:mm, j in 1:nn, k in 1:pp
    matrix1 = zeros(Int, mm, nn)
    matrix2 = zeros(Int, nn, pp)
    matrix3 = zeros(Int, pp, mm)

    matrix1[i,j] = 1
    matrix2[j,k] = 1
    matrix3[k,i] = 1

    T1 = T1 + kron(kron(matrix1, matrix2), matrix3)
end

T1
```

```julia
# Exact symbolic verification of all 4096 Brent equations.
D = simplify_expr.(T .- T1)
nonzero_positions = findall(q -> !iszero(q), D)

println("number of nonzero entries in T-T1 = ", length(nonzero_positions))
```
